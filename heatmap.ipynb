```python

import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

sns.set_theme()

bins = [0,15,25,65,100]
category_names = ["Kinder", "Jugendliche", "Erwachsene", "Senioren"]
df["Age_Group"] = pd.cut(df["Age"], bins=bins, labels=category_names)

heatmap_data = df.pivot_table(
    index="Age_Group",
    columns="Sex",
    values="Survived",
    aggfunc="count"
)

ax = sns.heatmap(heatmap_data, cmap="Greens")
ax.invert_yaxis()
```
